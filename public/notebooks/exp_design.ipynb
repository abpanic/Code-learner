# A/B experiment decision notebook

Use this as a small template for a pre-registered experiment: specify the **unit of randomization**, primary metric, guardrails and minimum detectable effect before looking at results. This five-per-arm example illustrates arithmetic only; it is too small for a credible launch decision.


```python
control = [10, 12, 9, 11, 10]
treatment = [12, 12, 11, 13, 12]
mean = lambda values: sum(values) / len(values)
a, b = mean(control), mean(treatment)
print(f"control={a:.1f}, treatment={b:.1f}")
print(f"absolute lift={b-a:.1f}; relative lift={(b-a)/a:.1%}")
```

## Before making a decision

1. Verify exposure and sample-ratio balance.
2. Estimate uncertainty at the **randomization unit**, accounting for repeated observations.
3. Check guardrails, segments and the pre-specified stopping rule.
4. Compare the expected business value against rollout cost and risk.
